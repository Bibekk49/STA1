## Overall conclusion

> Write approximately 150–250 words explaining how sample size and sampling variability affect
> sample means, standard errors, and confidence-interval precision. Report the two parameter
> estimates with uncertainty and state one design limitation for each study.

**Sample size and variability:** a sample mean changes from sample to sample, and its standard
error $\sigma/\sqrt n$ measures how much. In Part 1 the SE of the mean restoration time fell from
60 minutes for one observation to 6.71 minutes for 80. Because the SE shrinks with $\sqrt n$,
halving it takes four times as many observations. A confidence interval has the basic form
estimate ± critical value × SE, so a smaller SE gives a narrower, more precise interval, while more
variable data or a higher confidence level make it wider. For a margin of 0.02 in Part 3, 734
attempts would be needed, or 2401 without a pilot estimate of $p$. More data reduce only random
error; they do not fix bias, dependence or an unrepresentative sample.

**Estimates:** the mean calibration error $\mu$ is estimated as $\bar x = 0.19$ °C, with 95 % $t$
interval $[0.03,\ 0.34]$ °C from 64 sensors. The false-rejection probability $p$ is estimated as
$\hat p = 0.083$, with 95 % Wilson interval $[0.055,\ 0.125]$ from 240 attempts.

**Design limitations:** every calibration error is measured against the reference temperature, so a
biased reference would shift all 64 errors, and more sensors would not reveal it. The login attempts
were staged under test conditions, so real users with other devices and habits may be falsely
rejected at a different rate.
